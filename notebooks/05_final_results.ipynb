## Final Model Interpretation

The final hybrid model combines TF-IDF text features with structured job-advertisement metadata using Logistic Regression. A classification threshold of 0.85 was selected using the validation set to obtain a better balance between fraud precision and recall.

On the untouched random test set, the model achieved:

* **Accuracy:** 98.38%
* **Fraud Precision:** 95.28%
* **Fraud Recall:** 69.94%
* **Fraud F1-Score:** 80.67%
* **ROC-AUC:** 99.12%
* **PR-AUC:** 92.30%

The high fraud precision indicates that most advertisements flagged as fraudulent were actually fraudulent. The fraud recall of 69.94% shows that the model successfully identified a substantial proportion of fraudulent advertisements, although some fraudulent advertisements were still missed.

The group-aware evaluation produced lower performance than the random split, demonstrating that duplicate or repeated job-description patterns can make random-split evaluation optimistic. This highlights the importance of leakage-resistant evaluation for fraudulent job advertisement detection.

Overall, the results indicate that combining textual information with structured metadata can provide an effective approach for detecting potentially fraudulent job advertisements. The model can serve as a screening system to prioritize suspicious advertisements for further human investigation rather than replacing human verification.
